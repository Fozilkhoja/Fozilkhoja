```python
total = len(teen)
female = len(teen[teen['gender']=='female'])
male = len(teen[teen['gender']=='male'])

pct_f = (female/total)*100
pct_m = (male/total)*100

print(f"Yigitlar: {male} ta ({pct_m:.1f}%)")
print(f"Qizlar: {female} ta ({pct_f:.1f}%)")
```

```python
#O'smirlar eng ko'p qaysi platformadan (platform_usage) foydalanishadi? Har bir platformadan foydalanuvchilar soni

teen['platform_usage'].value_counts()
```

```python
#Uxlashdan oldin ekran vaqti 2 soatdan ko'p bo'lgan o'smirlarning necha foizi o'qishda (academic_performance) past ko'rsatkichga ega?
screen = teen[teen['screen_time_before_sleep'] > 2]
screen['screen_time_before_sleep'].value_counts()
pct = (screen['academic_performance'] < 3).mean()*100

print(f"Foizda: {pct:.2f}%")
```

```python
#depression_label ustunida 1 (bor) deb belgilangan o'smirlarning o'rtacha ijtimoiy tarmoqda o'tirish soati umumiy o'smirlarnikidan
#qanchalik farq qiladi?
dep = teen[teen['depression_label'] == 1]['daily_social_media_hours'].mean()
total_avg = teen['daily_social_media_hours'].mean()
dif = dep - total_avg

print(f"Depression_labeli 1 ga teng o'smirlar o'rtacha {dif} soatga ko'proq ijtimoiy tarmoqda o'tirishadi.")
```

```python
#yangi sleep_quality (uyqu sifati) ustunini oching: agar sleep_hours 7 dan ko'p bo'lsa 'Yaxshi', 7 dan kam bo'lsa 'Yomon' deb qiymat bering.
teen['sleep_quality'] = np.where(teen['sleep_hours']>7, 'Yaxshi', 'Yomon')
teen.drop('quality', axis=1, inplace=True)
teen.head()
```

Har bir yosh bo'yicha o'smirlarning o'rtacha stress (stress_level), xavotir (anxiety_level) va o'qish ko'rsatkichini (academic_performance) hisoblang. Qaysi yoshda stress eng yuqori?

```python
teen.groupby('age').agg({'stress_level': 'mean', 'anxiety_level': 'mean', 'academic_performance': 'mean'})
```

```python
#Jins va platforma kesimida o'smirlarning o'rtacha ijtimoiy tarmoqda o'tkazadigan soatlarini hisoblang.
pivot = teen.pivot_table(
    columns='gender',
    index='platform_usage',
    values='daily_social_media_hours',
    aggfunc='mean')
pivot
```

```python
#Ijtimoiy aloqa darajasi bo'yicha o'smirlarning o'rtacha telefonga qaramlik darajasini aniqlang.

teen.groupby('social_interaction_level')['addiction_level'].mean()
```

```python
#Sonli ustunlar o'rtasidagi bog'liqlikni toping. Qaysi ikki omil bir-biriga eng kuchli bog'langan (musbat yoki manfiy)?
teen.corr(numeric_only=True)
```

```python
#"Jismoniy faollik stressni kamaytiradi" degan fikr to'g'rimi?
#Sport bilan haftasiga 1 soatdan ko'p shug'ullanadiganlar va umuman shug'ullanmaydiganlarning o'rtacha stress darajasini solishtiring.

no_sport = teen[teen['physical_activity']==0]
stress = no_sport['stress_level'].mean()

sport = teen[teen['physical_activity']>1]
stress_s = sport['stress_level'].mean()

print(f"Sport bilan shug'ullanmaydiganlarning o'rtacha stress darajasi: {stress}")
print(f"Sport bilan shug'ullanadiganlarning o'rtacha stress darajasi: {stress_s}")
```
